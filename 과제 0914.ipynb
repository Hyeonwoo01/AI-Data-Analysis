# #1

```python
from sklearn.datasets import make_regression
import random
import numpy as np

data = make_regression(n_samples=1000, n_features=1, bias=200, noise=30, random_state=42)
X_raw = data[0]
y_raw = data[1]

epochs = 1000
learning_rate = 0.01
n_data = len(X_raw)
```

## 기본 파이썬

```python
X_list = [x[0] for x in X_raw]
y_list = list(y_raw)

w1 = random.random()
w0 = random.random()

for epoch in range(epochs):
    total_error = 0
    dw1_sum = 0
    dw0_sum = 0

    for i in range(n_data):
        x_i = X_list[i]
        y_i = y_list[i]
        
        y_pred_i = w1 * x_i + w0
        
        error = y_pred_i - y_i
        total_error += error ** 2
        
        dw1_sum += 2 * error * x_i
        dw0_sum += 2 * error
        
    cost = total_error / n_data
    dw1 = dw1_sum / n_data
    dw0 = dw0_sum / n_data
    
    w1 = w1 - learning_rate * dw1
    w0 = w0 - learning_rate * dw0

print(f"Cost: {cost:.4f}")
print(f"가중치 w1: {w1_np:.4f}")
print(f"편향 w0: {w0_np:.4f}")
```

## Numpy

```python
X_np = X_raw.ravel() 
y_np = y_raw

w1_np = np.random.rand()
w0_np = np.random.rand()

for epoch in range(epochs):
    y_pred = w1_np * X_np + w0_np
    
    cost = np.sum((y_pred - y_np) ** 2) / n_data
    
    dw1 = 2 * np.dot((y_pred - y_np), X_np) / n_data
    dw0 = 2 * np.mean(y_pred - y_np)
    
    w1_np = w1_np - learning_rate * dw1
    w0_np = w0_np - learning_rate * dw0


print(f"Cost: {cost:.4f}")
print(f"가중치 w1: {w1_np:.4f}")
print(f"편향 w0: {w0_np:.4f}")
```

# #2

```python
import numpy as np
import random
from sklearn.datasets import make_regression

data = make_regression(n_samples=5000, n_features=100, bias=200, noise=50, random_state=42)
X_raw = data[0]
y_raw = data[1]

n_data = 5000
n_features = 100
epochs = 100          
learning_rate = 0.1   
```

## 기본 파이썬

```python
X_list = X_raw.tolist()
y_list = y_raw.tolist()

w = [random.random() for _ in range(n_features)]
b = random.random()

for epoch in range(epochs):
    dw = [0] * n_features 
    db = 0
    total_cost = 0
    
    for i in range(n_data):
        y_pred_i = sum(w[j] * X_list[i][j] for j in range(n_features)) + b
        
        error = y_pred_i - y_list[i]
        total_cost += error ** 2
        
        db += 2 * error
        for j in range(n_features):
            dw[j] += 2 * error * X_list[i][j]
            
    b -= learning_rate * (db / n_data)
    for j in range(n_features):
        w[j] -= learning_rate * (dw[j] / n_data)
cost = total_cost / n_data

print(f"Cost: {cost:.4f}")
print(f"전체 가중치 w: {w_np}")  
print(f"편향 b: {b_np:.4f}")      
```

## Numpy

```python
w_np = np.random.rand(n_features)
b_np = np.random.rand()

for epoch in range(epochs):
    y_pred = X_raw @ w_np + b_np
    
    error = y_pred - y_raw
    cost = np.sum(error ** 2) / n_data
    
    dw_np = 2 * (X_raw.T @ error) / n_data
    db_np = 2 * np.mean(error)
    
    w_np -= learning_rate * dw_np
    b_np -= learning_rate * db_np

print(f"Cost: {cost:.4f}")
print(f"전체 가중치 w: {w_np}")
print(f"편향 b: {b_np:.4f}")
```
