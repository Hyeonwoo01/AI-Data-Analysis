```python
def get_middle_char(text):
    length = len(text)
    #print(length)
    mid = length // 2  
    #print(mid)

    # 글자수: 홀수
    if length % 2 == 1:
        return text[mid]
    
    # 글자수: 짝수
    else:
        return text[mid - 1 : mid + 1]

print(get_middle_char('study'))   # 결과: u
print(get_middle_char('python'))  # 결과: th
```

```python
def my_avg(*args):
    return sum(args) / len(args)

print(my_avg(77, 83, 95, 80, 70))  # 결과: 81.0
```
