```python
# 상권 데이터 로드 및 시군구 단위 집계
df_store = pd.read_csv('../data/raw/소상공인시장진흥공단_상가(상권)정보_서울_202606.csv', encoding='utf-8-sig', encoding_errors='replace')
df_store['시군구명'] = df_store['시군구명'].apply(lambda x: str(x).strip())
df_store_agg = df_store.groupby('시군구명').size().reset_index(name='총상가수')

# 인구 데이터 로드
df_pop = pd.read_csv('../data/raw/행정안전부_지역별&#40;행정동&#41; 성별 연령별 주민등록 인구수_20260630.csv', encoding='utf-8-sig', encoding_errors='replace')

df_pop['시군구명'] = df_pop['시군구명'].apply(lambda x: str(x).strip())


age_20s = [f'{i}세남자' for i in range(20, 30)] + [f'{i}세여자' for i in range(20, 30)]
df_pop['20대인구수'] = df_pop[age_20s].sum(axis=1)

age_30s = [f'{i}세남자' for i in range(30, 40)] + [f'{i}세여자' for i in range(30, 40)]
df_pop['30대인구수'] = df_pop[age_30s].sum(axis=1)

age_65_m = [c for c in df_pop.columns if '세남자' in c and int(c.replace('세남자', '')) >= 65]
age_65_f = [c for c in df_pop.columns if '세여자' in c and int(c.replace('세여자', '')) >= 65]

age_110 = ['110세이상 남자', '110세이상 여자']
age_65_plus = age_65_m + age_65_f + [c for c in age_110 if c in df_pop.columns]
df_pop['65세이상인구수'] = df_pop[age_65_plus].sum(axis=1)

# 시군구별 인구 통계 합산
df_pop_agg = df_pop.groupby('시군구명')[['계', '20대인구수', '30대인구수', '65세이상인구수']].sum().reset_index()

# 시군구 기준으로 조인
df_merged = pd.merge(
    df_store_agg, 
    df_pop_agg, 
    on='시군구명', 
    how='inner'
)

print(f"병합된 시군구 수: {len(df_merged)}")
df_merged.head()
```

```python
import pandas as pd
import pymysql

df_store = pd.read_csv('../data/raw/소상공인시장진흥공단_상가(상권)정보_서울_202606.csv', encoding='utf-8-sig', encoding_errors='replace')
df_pop = pd.read_csv('../data/raw/행정안전부_지역별&#40;행정동&#41; 성별 연령별 주민등록 인구수_20260630.csv', encoding='utf-8-sig', encoding_errors='replace')

store_cols = {
    '상가업소번호': 'store_id', '상호명': 'store_name',
    '상권업종대분류코드': 'category_large_code', '상권업종대분류명': 'category_large',
    '상권업종중분류코드': 'category_mid_code', '상권업종중분류명': 'category_mid',
    '시군구명': 'sigungu', '행정동코드': 'region_code',
    '행정동명': 'region_name', '경도': 'lon', '위도': 'lat'
}
df_store_new = df_store[list(store_cols.keys())].rename(columns=store_cols)
df_store_new = df_store_new.groupby('store_id').head(1)
df_store_new = df_store_new.dropna(subset=['store_id', 'sigungu'])
df_store_new['region_code'] = df_store_new['region_code'].apply(lambda x: str(x).strip())
df_store_new['sigungu'] = df_store_new['sigungu'].apply(lambda x: str(x).strip())

df_pop['시군구명'] = df_pop['시군구명'].apply(lambda x: str(x).strip())

age_groups = [0, 10, 20, 30, 40, 50, 60, 70]
for age in age_groups:
    if age == 70:
        cols_m = [c for c in df_pop.columns if '세남자' in c and int(c.replace('세남자', '')) >= 70]
        cols_f = [c for c in df_pop.columns if '세여자' in c and int(c.replace('세여자', '')) >= 70]
        cols_plus = cols_m + cols_f + [c for c in ['110세이상 남자', '110세이상 여자'] if c in df_pop.columns]
        df_pop[f'pop_{age}s'] = df_pop[cols_plus].sum(axis=1)
    else:
        cols_m = [f'{i}세남자' for i in range(age, age+10) if f'{i}세남자' in df_pop.columns]
        cols_f = [f'{i}세여자' for i in range(age, age+10) if f'{i}세여자' in df_pop.columns]
        df_pop[f'pop_{age}s'] = df_pop[cols_m + cols_f].sum(axis=1)

age_65_m = [c for c in df_pop.columns if '세남자' in c and int(c.replace('세남자', '')) >= 65]
age_65_f = [c for c in df_pop.columns if '세여자' in c and int(c.replace('세여자', '')) >= 65]
age_65_plus = age_65_m + age_65_f + [c for c in ['110세이상 남자', '110세이상 여자'] if c in df_pop.columns]
df_pop['pop_65plus'] = df_pop[age_65_plus].sum(axis=1)

pop_agg_cols = ['계', 'pop_0s', 'pop_10s', 'pop_20s', 'pop_30s', 'pop_40s', 'pop_50s', 'pop_60s', 'pop_70s', 'pop_65plus']
df_pop_new = df_pop.groupby('시군구명')[pop_agg_cols].sum().reset_index()
df_pop_new.columns = ['sigungu', 'total_pop', 'pop_0s', 'pop_10s', 'pop_20s', 'pop_30s', 'pop_40s', 'pop_50s', 'pop_60s', 'pop_70s', 'pop_65plus']

df_store_new = df_store_new.astype(object).where(pd.notnull(df_store_new), None)
df_pop_new = df_pop_new.astype(object).where(pd.notnull(df_pop_new), None)

conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='test1234',  
    database='project_db',
    port=3306,
    charset='utf8mb4'
)

try:
    with conn.cursor() as cur:
        with open('../db/schema.sql', 'r', encoding='utf-8') as f:
            sql_queries = f.read().split(';')
            for query in sql_queries:
                if query.strip():
                    cur.execute(query)
        print("스키마 적용 완료")

        # 상권 데이터 적재
        store_sql = """
            INSERT INTO store_sales 
            (store_id, store_name, category_large_code, category_large, category_mid_code, category_mid, sigungu, region_code, region_name, lon, lat) 
            VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
        """
        store_data = [tuple(x) for x in df_store_new.to_numpy()]
        cur.executemany(store_sql, store_data)

        # 인구 데이터 적재
        pop_sql = """
            INSERT INTO population 
            (sigungu, total_pop, pop_0s, pop_10s, pop_20s, pop_30s, pop_40s, pop_50s, pop_60s, pop_70s, pop_65plus) 
            VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
        """
        pop_data = [tuple(x) for x in df_pop_new.to_numpy()]
        cur.executemany(pop_sql, pop_data)

    conn.commit()
    print("데이터 적재 완료")

except Exception as e:
    conn.rollback()
    print(f"[에러 발생] 데이터 적재 실패: {e}")

finally:
    with conn.cursor() as cur:
        cur.execute("SELECT COUNT(*) FROM store_sales")
        store_count_db = cur.fetchone()[0]

        cur.execute("SELECT COUNT(*) FROM population")
        pop_count_db = cur.fetchone()[0]

        print(f"\n[검증 결과]")
        print(f"상권 원본 행 수: {len(df_store_new)} / DB 적재 행 수: {store_count_db}")
        print(f"인구 원본 행 수: {len(df_pop_new)} / DB 적재 행 수: {pop_count_db}")
        
    conn.close()
```

```python
conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='test1234',  
    database='project_db',
    port=3306,
    charset='utf8mb4'
)

query2 = """
/* [가설 2: 청년 인구와 선택적 소비 업종의 관계]를 검증하기 위해,
   WITH와 CASE WHEN를 사용하여 주점/오락/스포츠 업종의 비중을 산출합니다. */
WITH YouthStore AS (
    SELECT 
        sigungu, 
        COUNT(store_id) AS total_stores,
        SUM(CASE WHEN category_mid IN ('주점', '오락용품 소매', '스포츠 서비스', '비알코올 음료점업') THEN 1 ELSE 0 END) AS youth_stores
    FROM store_sales
    GROUP BY sigungu
)
SELECT 
    y.sigungu, 
    (y.youth_stores * 100.0 / y.total_stores) AS youth_store_ratio,
    ((p.pop_20s + p.pop_30s) * 100.0 / p.total_pop) AS youth_pop_ratio
FROM YouthStore y
JOIN population p ON y.sigungu = p.sigungu
LIMIT 5;
"""
print(pd.read_sql(query2, con=conn))

conn.close()
```

```python
import pandas as pd
import pymysql
import matplotlib.pyplot as plt
import seaborn as sns

conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='test1234',  
    database='project_db',
    port=3306,
    charset='utf8mb4'
)

df_store = pd.read_sql("SELECT * FROM store_sales", con=conn)
df_pop = pd.read_sql("SELECT * FROM population", con=conn)

conn.close()

def check_youth_keyword(category):
    if type(category) != str: 
        return False

    keywords = ['카페', '커피', '주점', '호프', '오락', '스포츠', 'PC방', '노래방']
    for keyword in keywords:
        if keyword in category:
            return True
    return False

# 키워드 확인
bool_indexes = df_store['category_mid'].apply(check_youth_keyword)
df_youth_store = df_store[bool_indexes]

df_youth_cnt = df_youth_store.groupby('sigungu').size().reset_index(name='youth_stores')

# 총 상가 수 집계
df_store_cnt = df_store.groupby('sigungu').size().reset_index(name='total_stores')

# 데이터 병합 및 비율 계산
df_h2 = pd.merge(df_store_cnt, df_youth_cnt, on='sigungu', how='left').fillna(0)
df_h2['youth_store_ratio'] = (df_h2['youth_stores'] / df_h2['total_stores']) * 100

df_h2 = pd.merge(df_h2, df_pop, on='sigungu', how='inner')
df_h2['youth_pop_ratio'] = ((df_h2['pop_20s'] + df_h2['pop_30s']) / df_h2['total_pop']) * 100

# 시각화
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(9, 6))
sns.scatterplot(data=df_h2, x='youth_pop_ratio', y='youth_store_ratio', s=120, color='darkorange')
sns.regplot(data=df_h2, x='youth_pop_ratio', y='youth_store_ratio', scatter=False, color='red')

plt.title("[가설 2] 청년층(2030) 인구 비율과 청년 선호 업종 비중", fontsize=13, fontweight='bold')
plt.xlabel('20~30대 인구 비율 (%)', fontsize=11)
plt.ylabel('전체 상가 중 청년 선호 업종 비중 (%)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

print(f"\n[가설 2 상관계수]")
print(f"청년 인구 비율과 청년 선호 업종 비중의 상관계수: {df_h2['youth_pop_ratio'].corr(df_h2['youth_store_ratio']):.4f}")
```

해석: 청년들이 많이 거주하는 구라고 해서, 해당 지역에 청년 선호 상가가 더 많이 집중되어 있는 것이 아닌 것을 알 수 있습니다.

인사이트/한계: 특정 구에 청년들이 많이 거주하는 것이 그 지역의 청년 상권을 발달시키는 것이라는 직접적인 인과관계가 성립하지 않음을 알 수 있습니다. 청년층은 거주지 주변에서 소비하기보다 교통이 발달한 외부로 이동하는 성향이 강하므로, 거주 인구 중심의 분석만으로는 타겟 상권의 입지를 예측할 수 없음을 알 수 있습니다.

시각화 목적:

    - 2030 청년 인구 거주 비율과 청년 선호 업종 상가 비중을 산점도와 추세선으로 시각화하여, 청년이 많이 사는 곳에 청년 상권이 발달할 것이라는 일반적인 편견과 달리 실제로는 거주지와 소비 상권이 무슨 인과관계가 있는지 증명하기 위함입니다.

```python
import pandas as pd
import pymysql
import matplotlib.pyplot as plt
import seaborn as sns

conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='test1234',  
    database='project_db',
    port=3306,
    charset='utf8mb4'
)

df_store = pd.read_sql("SELECT * FROM store_sales", con=conn)
df_pop = pd.read_sql("SELECT * FROM population", con=conn)

conn.close()

def check_medical_keyword(category):
    if type(category) != str: 
        return False

    keywords = ['병원', '의원', '보건', '의약']
    for keyword in keywords:
        if keyword in category:
            return True
    return False

# 키워드 확인
medical_mask = df_store['category_mid'].apply(check_medical_keyword)
df_medical_store = df_store[medical_mask]

df_medical_cnt = df_medical_store.groupby('sigungu').size().reset_index(name='medical_stores')
df_store_cnt = df_store.groupby('sigungu').size().reset_index(name='total_stores')

# 데이터 병합 및 비율 계산
df_h3 = pd.merge(df_store_cnt, df_medical_cnt, on='sigungu', how='left').fillna(0)
df_h3['medical_store_ratio'] = (df_h3['medical_stores'] / df_h3['total_stores']) * 100

df_h3 = pd.merge(df_h3, df_pop, on='sigungu', how='inner')
df_h3['elderly_pop_ratio'] = (df_h3['pop_65plus'] / df_h3['total_pop']) * 100

# 시각화
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(9, 6))
sns.scatterplot(data=df_h3, x='elderly_pop_ratio', y='medical_store_ratio', s=120, color='teal')
sns.regplot(data=df_h3, x='elderly_pop_ratio', y='medical_store_ratio', scatter=False, color='red')

plt.title("[가설 3] 고령 인구(65세 이상) 비율과 의료/보건 업종 비중", fontsize=14, fontweight='bold')
plt.xlabel('65세 이상 고령 인구 비율 (%)', fontsize=12)
plt.ylabel('전체 상가 중 의료/보건 비중 (%)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 상관계수 출력
print(f"\n[가설 3 상관계수 확인]")
print(f"65세 이상 인구 비율과 의료/보건 업종 비중의 상관계수: {df_h3['elderly_pop_ratio'].corr(df_h3['medical_store_ratio']):.4f}")
```

해석: 고령화 지수가 높은 지역이라고 해서 전체 상권 중 병원/의원 등의 의료 인프라 비중이 뚜렷하게 높아지지는 않습니다.

인사이트/한계: 병원과 약국 등은 고령층의 수요가 높지만, 동시에 전 연령대가 이용하는 필수 인프라입니다. 따라서 특정 지역의 고령화 지수가 높아진다고 해서 그 동네에 병원이 뚜렷하게 많아질 것이라는 단순한 인과관계로 해석할 수 없습니다. 특히 대형 병원일수록 거주 인구의 연령대보다는 광역 교통망 중심의 입지를 선호한다는 사실을 알 수 있습니다.

 시각화 목적:
 
    - 65세 이상 고령층 거주 비율과 의료/보건 업종 상가 비중의 관계를 시각화하여, 필수 인프라인 병원/약국이 단순히 고령층 거주 비율에 비례해서 생길 것이라는 일반적인 편견과 달리 광역 교통망 등 다른 요인에 의해 분포하고 있음을 보여주기 위함입니다.


```python
conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='test1234',  
    database='project_db',
    port=3306,
    charset='utf8mb4'
)

query4 = """
/* 전체 구의 평균 상가 수를 초과하는 
   상권 밀집 구역만 서브쿼리로 필터링하여 인구 특성을 조회합니다. */
SELECT 
    p.sigungu, 
    p.total_pop, 
    p.pop_20s, 
    p.pop_65plus
FROM population p
WHERE p.sigungu IN (
    SELECT sigungu 
    FROM store_sales 
    GROUP BY sigungu 
    HAVING COUNT(store_id) > (SELECT COUNT(*)/25 FROM store_sales)
)
LIMIT 5;
"""

print(pd.read_sql(query4, con=conn))

conn.close()
```

```python
conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='test1234',  
    database='project_db',
    port=3306,
    charset='utf8mb4'
)

query5 = """
/* [가설 4: 상권 업종 쏠림 현상] 분석을 보조하기 위해,
   각 시군구별로 가장 점포 수가 많은 1위 업종 카테고리가 무엇인지 도출합니다. */
WITH CategoryCount AS (
    SELECT sigungu, category_mid, COUNT(store_id) as cnt
    FROM store_sales
    GROUP BY sigungu, category_mid
),
MaxCategory AS (
    SELECT sigungu, MAX(cnt) as max_cnt
    FROM CategoryCount
    GROUP BY sigungu
)
SELECT c.sigungu, c.category_mid AS top_category, c.cnt AS store_count
FROM CategoryCount c
JOIN MaxCategory m ON c.sigungu = m.sigungu AND c.cnt = m.max_cnt
ORDER BY c.sigungu
LIMIT 5;
"""
print(pd.read_sql(query5, con=conn))

conn.close()
```

```python
import pandas as pd
import pymysql
import matplotlib.pyplot as plt
import seaborn as sns

conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='test1234',  
    database='project_db',
    port=3306,
    charset='utf8mb4'
)

df_store = pd.read_sql("SELECT sigungu, category_mid FROM store_sales", con=conn)
df_pop_db = pd.read_sql("SELECT sigungu, pop_0s, pop_10s, pop_20s, pop_30s, pop_40s, pop_50s, pop_60s, pop_70s FROM population", con=conn)

conn.close()

# 상권 업종 분산 계산
df_store['count'] = 1
store_counts = df_store.pivot_table(index='sigungu', columns='category_mid', values='count', aggfunc='sum').fillna(0)

store_prop = store_counts.copy()
total_per_sigungu = store_counts.sum(axis=1) 
for col in store_prop.columns:
    store_prop[col] = store_prop[col] / total_per_sigungu

store_variance = store_prop.std(axis=1).reset_index(name='store_std')
store_variance['store_concentration'] = store_variance['store_std'] * store_variance['store_std']
store_variance.drop(['store_std'], axis=1, inplace=True)

# 인구 연령대 분산 계산
df_pop_agg = df_pop_db.set_index('sigungu') 

pop_prop = df_pop_agg.copy()
total_pop_per_sigungu = df_pop_agg.sum(axis=1)
for col in pop_prop.columns:
    pop_prop[col] = pop_prop[col] / total_pop_per_sigungu

pop_variance = pop_prop.std(axis=1).reset_index(name='pop_std')
pop_variance['age_concentration'] = pop_variance['pop_std'] * pop_variance['pop_std']
pop_variance.drop(['pop_std'], axis=1, inplace=True)


# 시각화
df_h4 = pd.merge(pop_variance, store_variance, on='sigungu', how='inner')

plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(9, 6))
sns.scatterplot(data=df_h4, x='age_concentration', y='store_concentration', s=120, color='purple')
sns.regplot(data=df_h4, x='age_concentration', y='store_concentration', scatter=False, color='red')

plt.title("[가설 4] 인구 연령 쏠림 현상과 상권 업종 쏠림 관계", fontsize=14, fontweight='bold')
plt.xlabel('연령 분포 쏠림 정도', fontsize=12)
plt.ylabel('상권 업종 쏠림 정도', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# 상관계수 확인
print(f"\n[가설 4 상관계수 확인]")
print(f"연령 분포 쏠림과 상권 업종 쏠림의 상관계수: {df_h4['age_concentration'].corr(df_h4['store_concentration']):.4f}")
```

해석: 주민들의 연령대가 한쪽으로 크게 쏠려 있는 지역이라도, 상권의 업종까지 특정 카테고리로 획일화되는 현상은 발생하지 않았습니다.

인사이트/한계: 상권의 업종 다양성을 결정짓는 진짜 원인은 거주민의 연령 구조보다 임대료, 상업 지구의 용도 및 특성, 외부 관광객이나 직장인 유동인구의 수요 등 훨씬 복합적인 외부 환경 요인들에 의해 결정된다고 볼 수 있습니다.

시각화 목적:

    - 각 자치구의 연령 분포 쏠림 정도와 상권 업종 쏠림 정도를 시각화하여, 동네에 특정 연령대만 산다고 해서 동네 상권까지 획일화되지는 않으며, 상권의 다양성은 거주민 연령 외에 훨씬 복합적인 요인이 작용함을 확인하기 위함입니다.
