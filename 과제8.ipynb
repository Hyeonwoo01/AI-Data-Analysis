```python
result_lines = ["name,average\n"]


with open('scores.csv', 'r', encoding='utf-8') as f:
    lines = f.readlines() 

header = lines[0].rstrip().split(',') 
subjects = header[1:]

# 학생별 데이터 처리
for line in lines[1:]:
    data = line.rstrip('\n').split(',') #
    name = data[0]
    scores = []
    
    for i in range(1, 4):
        subject_name = subjects[i-1]
        value = data[i]
        
        try:
            score = float(value)
            scores.append(score)
        except ValueError: 
            print(f"[처리] {name} - {subject_name} 값 오류 ({value}) -> 0점 처리")
            scores.append(0.0) 
            
    # 평균 계산 및 반올림 (첫째 자리까지)
    avg = round(sum(scores) / len(scores), 1)
    
    result_lines.append(f"{name},{avg}\n")

with open('result.csv', 'w', encoding='utf-8') as f:
    f.writelines(result_lines)

print("완료: result.csv 생성")
```

```python
import json

success_count = 0
fail_count = 0
valid_members = []
error_logs = []

with open('members.csv', 'r', encoding='utf-8') as f:
    lines = f.readlines()

for line_num, line in enumerate(lines, start=1):
    if line_num == 1:
        continue
        
    data = line.rstrip('\n').split(',') 
    
    row_id = data[0]
    name = data[1]
    age_str = data[2]

    if not name.strip():
        error_logs.append(f"{line_num}행: name 비어있음\n")
        fail_count += 1
        continue

    try:
        member_dict = {
            "id": int(row_id),
            "name": name,
            "age": int(age_str)
        }
   
        valid_members.append(member_dict)
        success_count += 1
        
    except ValueError as e:
        error_logs.append(f"{line_num}행: {e}\n")
        fail_count += 1

with open('members.json', 'w', encoding='utf-8') as f:
    json.dump(valid_members, f, ensure_ascii=False, indent=2)

if error_logs:
    with open('errors.log', 'w', encoding='utf-8') as f:
        f.writelines(error_logs) 

print(f"성공 {success_count}건 / 실패 {fail_count}건")
```
