```python
scores = {
    'python': 80,
    'algorithm': 90,
    'django': 89,
    'web': 83,
} # => 85.5
total_score = sum(scores.values())
subject = len(scores)
average = total_score / subject
print(average)
```

```python
bloods = [
    'A', 'B', 'A', 'O', 'AB', 'AB',
    'O', 'A', 'B', 'O', 'B', 'AB',
] # => {'A': 3, 'B': 3, 'O':  3, 'AB': 3}
blood_count = {}

for blood in bloods:
    if blood in blood_count:
        blood_count[blood] += 1
    else:
        blood_count[blood] = 1
print(blood_count)
```
