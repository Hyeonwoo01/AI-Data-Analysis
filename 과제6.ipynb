```python
#1. 
#장점: 코드가 더 직관적이고 이해하기 쉬운 경우가 있습니다.
#     알고리즘을 설계하고 구현하는 데 유용하게 활용됩니다.

#단점: 함수가 호출될 때마다 메모리 공간에 데이터가 지속해서 쌓이게 됩니다.
#     이로 인해 메모리 스택이 넘치는 현상(Stack overflow)이 발생할 위험이 있습니다.
#     프로그램의 실행 속도가 느려지는 단점이 생깁니다.


# 재귀 호출은 입력 값이 커질 수록 연산 속도가 오래걸립니다.
# 반복문은 재귀로 구현된 함수보다 연산 속도가 빠른 편입니다.
# 알고리즘 자체가 재귀적인 표현이 자연스러운 경우 재귀함수를 사용합니다.
# 재귀 호출은 `변수 사용` 을 줄여줄 수 있습니다.


```

```python
#2
class Circle:
    pi = 3.14

    def __init__(self, r, x, y):
        self.r = r
        self.x = x
        self.y = y

    def area(self):
        # 원의 넓이
        return Circle.pi * (self.r ** 2)

    def circumference(self):
        # 원의 둘레
        return 2 * Circle.pi * self.r

    def center(self):
        # 중심 좌표
        return (self.x, self.y)

c1 = Circle(3, 2, 4)

print(c1.area())          # 출력: 28.26
print(c1.circumference()) # 출력: 18.84
print(c1.center())        # 출력: (2, 4)
```
