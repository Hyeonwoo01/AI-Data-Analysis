### #1 ELMo 논문 리뷰

**1. Contextual Word Representations**

* 기존의 Word2Vec이나 GloVe 같은 전통적인 워드 임베딩 방식(정적 임베딩)은 문맥과 관계없이 하나의 단어에 무조건 하나의 고정된 벡터만 할당했습니다. 이 때문에 다의어나 동음이의어(예: 타는 '배', 먹는 '배')의 의미를 구분하지 못하는 치명적인 한계가 있었습니다.
* ELMo는 이를 극복하기 위해 단어가 사용된 주변 문맥을 파악하여 동적으로 벡터 값을 부여하는 **문맥적 단어 임베딩**을 제안했습니다. 따라서 문장 내 다른 단어들과의 관계를 고려하므로, 완전히 똑같은 단어라도 문맥에 따라 서로 다른 형태의 벡터로 정밀하게 표현될 수 있습니다.

**2. Bidirectional LSTM**

* ELMo는 문맥을 제대로 파악하기 위해 대규모 텍스트 코퍼스로 사전 학습된 양방향 언어 모델(biLM, Bidirectional Language Model)을 사용하는데, 이 모델의 핵심 뼈대가 바로 **양방향 LSTM(Bidirectional LSTM)** 구조입니다.
* 단순히 왼쪽에서 오른쪽으로만 텍스트를 읽는 것이 아니라, 순방향 LSTM으로 단어 이전의 문맥을 읽고, 역방향 LSTM으로 단어 이후의 문맥을 동시에 읽어 들입니다. 이 두 방향의 은닉 상태를 결합하여 단어 앞뒤의 전체적인 문맥 정보를 놓치지 않고 깊이 있게 학습합니다.

**3. Sentence-Level Embedding**

* ELMo는 단어 하나하나를 개별적으로 떼어서 처리하는 것이 아니라, 문장 전체를 모델의 입력으로 통째로 넣어 문장 단위의 문맥 흐름을 파악합니다.
* 또한, 양방향 LSTM의 여러 층에서 나온 출력값 중 최상위 층의 결과만 사용하는 것이 아니라, 모든 층의 은닉 상태 값들을 풀고자 하는 특정 자연어 처리 과제에 맞게 가중합하여 최종 벡터를 만듭니다. 하위 층은 문법적 특성을, 상위 층은 의미론적 특성을 잘 잡아내기 때문에 문장 전체의 풍부한 정보를 담아내는 임베딩을 완성할 수 있습니다.

### #2 Seq2Seq 논문 리뷰

**1. Encoder-Decoder 구조** 

* 이 논문은 길이가 서로 다른 가변 길이의 입력 시퀀스와 출력 시퀀스를 매핑할 수 있는 혁신적인 딥러닝 모델 구조를 제안했습니다. 모델은 크게 두 개의 다층 LSTM 네트워크로 구성됩니다. Encoder는 입력 시퀀스를 순차적으로 읽어 들여 그 전체적인 의미를 하나의 고정된 차원의 실수 벡터로 압축합니다. Decoder는 이 압축된 벡터를 넘겨받아 타겟 시퀀스의 단어들을 순차적으로 하나씩 예측하고 생성해 냅니다. 이 구조를 통해 입력과 출력의 차원이 고정되어야만 했던 기존 신경망 모델들의 치명적인 한계를 극복할 수 있었습니다.

**2. Machine Translation** 

* Seq2Seq 모델의 강력함이 가장 명확하게 입증된 응용 분야가 바로 기계 번역입니다. 논문에서는 영어 문장을 프랑스어 문장으로 번역하는 작업에 이 모델을 적용했습니다. 복잡한 언어학적 규칙이나 기존의 통계적 기계 번역 파이프라인에 의존하지 않고, 오직 거대한 인공 신경망 모델 자체만으로 원문을 번역문으로 바로 변환하는 엔드투엔드 학습을 수행하여 당시 SMT 시스템에 필적하는 매우 뛰어난 번역 성능을 달성했습니다.

**3. Sequence Transformation** 

* 이 논문은 단순히 번역 기술에 국한되지 않고, 임의의 시퀀스를 다른 임의의 시퀀스로 변환하는 매우 범용적이고 일반적인 프레임워크를 제시했습니다. 특히 논문에서 발견하여 적용한 가장 중요한 테크닉 중 하나는 **입력 시퀀스의 단어 순서를 거꾸로 뒤집어서 인코더에 입력**하는 것입니다. 문장을 뒤집어 입력하면 소스 문장의 시작 부분과 타겟 문장의 시작 부분 사이의 거리가 매우 가까워져 모델이 초반 예측을 수행하기 훨씬 쉬워집니다. 이는 RNN 계열 모델의 고질적인 한계인 '장기 의존성' 문제를 크게 완화시켜 주어 모델의 최적화 속도와 전체적인 예측 성능을 비약적으로 향상시키는 결과를 가져왔습니다.

### #3 Seq2Seq with Attention 논문 리뷰

**1. Attention Mechanism** 
* 기존의 Seq2Seq 모델은 인코더가 입력 문장을 하나의 고정 길이 벡터로 압축해야만 했기 때문에, 문장이 길어질수록 정보 손실이 발생하는 치명적인 병목 현상을 겪었습니다. 이 논문은 이를 해결하기 위해 어텐션이라는 혁신적인 메커니즘을 도입했습니다. 디코더가 타깃 단어를 매번 생성할 때마다 인코더의 모든 은닉 상태를 다시 한 번 살펴보고, 현재 번역해야 할 단어와 가장 연관성이 높은 소스 단어들에 더 많은 가중치를 부여하여 예측 성능을 비약적으로 끌어올렸습니다.

**2. Context Vector** 
* 기존 모델에서는 인코더의 마지막 은닉 상태만이 유일한 컨텍스트 벡터로 사용되었습니다. 반면 Attention 구조에서는 디코더가 단어를 생성하는 각 시점마다 서로 다른 동적인 컨텍스트 벡터를 새롭게 계산합니다. 이 벡터는 인코더의 각 은닉 상태와 계산된 어텐션 가중치들의 가중합으로 구해지며, 디코더가 매 순간 가장 필요로 하는 맞춤형 문맥 정보를 제공하는 역할을 합니다.

**3. Alignment and Translation** 
* 기존의 통계적 기계 번역 시스템에서는 소스 언어와 타깃 언어 간의 단어 매칭을 의미하는 정렬 과정을 번역 모델과 별도로 구축해야 했습니다. 하지만 이 논문에서 제안한 모델은 신경망 내부에 정렬 모델을 포함시켜, **정렬과 번역 작업을 하나로 통합하여 동시에 학습**합니다. 즉, 번역을 수행하면서 자연스럽게 '지금 소스 문장의 어떤 부분에 집중해야 하는가'를 모델 스스로 학습하고 최적화하게 됩니다.

**4. Variable-Length Input Handling** 
* 어텐션 메커니즘의 도입으로 얻게 된 가장 큰 성과 중 하나는 가변 길이의 긴 문장을 효과적으로 처리할 수 있게 되었다는 점입니다. 모든 정보를 하나의 고정된 벡터로 억지로 압축할 필요 없이, 입력 문장의 길이에 비례하여 생성된 여러 은닉 상태들을 유연하게 참조하기 때문에 **입력 시퀀스가 길어지더라도 정보 손실이 거의 발생하지 않으며, 긴 문장에서도 안정적이고 우수한 번역 성능을 유지**할 수 있게 되었습니다.
