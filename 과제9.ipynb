```python
import numpy as np
scores_list = [88, 92, 79, 95,
          75, 68, 82, 70,
          92, 99, 95, 88,
          60, 72, 65, 58,
          85, 80, 78, 90]
#1
print('1번 문제:')
scores = np.array(scores_list).reshape(5, 4)
print(f'shape: {scores.shape}')
print(f'ndim: {scores.ndim}')
print(f'size: {scores.size}')
print(f'dtype: {scores.dtype}\n')

#2
print('2번 문제:')
student_means = np.mean(scores, axis=1)
subject_means = np.mean(scores, axis=0)
print(f'학생 별 평균: {student_means}')
print(f'과목 별 평균: {subject_means}\n')

#3
print('3번 문제:')
pos = np.argmax(scores)
max_score = np.max(scores)
row = pos // 4
col = pos % 4
print(f'최고 점수: {max_score}')
print(f'1차원 인덱스 위치: {pos}')
print(f'2차원 인덱스 (행, 열): ({row}, {col}) -> {row + 1}번째 학생, {col + 1}번째 과목\n')

#4
print('4번 문제:')
total_mean = np.mean(scores)
total_std = np.std(scores)
total_var = np.var(scores)

print(f'전체 평균: {total_mean}')
print(f'전체 표준편차: {total_std:.2f}')
print(f'전체 분산: {total_var:.2f}\n')

#5
print('5번 문제:')
mask = student_means >= 80
indices = np.where(mask)[0]
count = np.sum(mask)

print(f"평균 80점 이상 학생 인덱스: {indices}")
print(f"평균 80점 이상 인원수: {count}명")
print("해당 학생들의 성적:\n", scores[mask], "\n")
```

```python
#6
print('6번 문제:')
count_90_above = np.sum(scores >= 90)
any_100 = np.any(scores >= 100)
all_60_above = np.all(scores >= 60)

print(f'90점 이상인 점수의 개수: {count_90_above}개')
print(f'100점 이상인 점수 여부: {any_100}')
print(f'모든 점수가 60점 이상 여부: {all_60_above}\n')

#7
print('7번 문제:')
grades = np.where(scores >= 90, 'A', np.where(scores >= 80, 'B', 'C'))
print('등급 배열:\n', grades)
```

#2

```python
np.random.seed(2024)
img = np.random.randint(0, 256, size=(8, 8))

#1
print('1번 문제:')
print(img, '\n')

#2
print('2번 문제:')
print(img[2:6, 2:6], '\n')

#3
print('3번 문제:')
print(img[::2], '\n')
```

```python
#4
print('4번 문제:')
bin_img = np.where(img >= 128, 255, 0)
print(bin_img, '\n')

#5
print('5번 문제:')
white_count = np.sum(bin_img == 255)
black_count = np.sum(bin_img == 0)
print(f'흰색 픽셀 개수: {white_count}개')
print(f'검은색 픽셀 개수: {black_count}개\n')
```

```python
#6
print('6번 문제:')
left, right = np.split(img, 2, axis=1)
print('좌측 절반:\n', left)
print('우측 절반:\n', right, '\n')
```

```python
#7
print('7번 문제:')
merged = np.concatenate([right, left], axis=1)
print(merged, '\n')

#8
print('8번 문제:')
flat = img.ravel()
print(f'앞 10개 값: {flat[:10]}')
print(f'shape: {flat.shape}\n')
#9
print('9번 문제:')
pick = img[img >= 128]
print(f'128 이상 픽셀 개수: {pick.size}')
print(f'128 이상 픽셀 평균: {np.mean(pick):.2f}')
```

```python
sales = np.array([[120,  80,  45,  60],
                  [200, 150,  90,  30],
                  [ 75,  60, 120,  95]])
price = np.array([1500, 3000, 5000, 2000])

# 1. 브로드캐스트 적용
revenue_matrix = sales * price
print('1. 브로드캐스트를 적용한 제품별 매출액:\n', revenue_matrix, '\n')
branch_total_sum = np.sum(revenue_matrix, axis=1)
print(branch_total_sum, '\n')

# 2. 행렬 연산(행렬곱) 적용
branch_total = sales.dot(price)
print('2. 행렬 연산을 적용한 지점별 총 매출액:\n', branch_total)
```
